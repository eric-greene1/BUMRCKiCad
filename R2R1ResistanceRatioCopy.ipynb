```python
# set the unversal variables

# Edit the commonResistors variable to contain whatever resistances are accessible
commonResistors = [10.0, 12.0, 15.0, 18.0, 22.0, 27.0, 33.0, 39.0, 47.0,
56.0, 68.0, 82.0, 100.0, 120.0, 150.0, 180.0, 220.0, 270.0,
330.0, 390.0, 470.0, 560.0, 680.0, 820.0, 1000.0, 1200.0,
1500.0, 1800.0, 2200.0, 2700.0, 3300.0, 3900.0, 4700.0,
5600.0, 6800.0, 8200.0, 10000.0, 12000.0, 15000.0, 18000.0,
22000.0, 27000.0, 33000.0, 47000.0, 68000.0, 100000.0,
220000.0, 330000.0, 470000.0, 1000000.0]

#Declare variables for later use
differences = []
idealRatios = [[]]
bestResistorIndicesR1 = []
bestResistorIndicesR2 = []
bestResistorsR1 = []
bestResistorsR2 = []
```

```python
# Create differencefinder function
def differenceFinder(desiredRatio, R2, R1):
    return abs(desiredRatio - (R2 / R1))

#Find the three indices which are the closest to the desired value within the differences list
def lowestThreeIndices(differences):
    sorted_indices = sorted(range(len(differences)), key=lambda i: differences[i])
    return sorted_indices[:3]
```

```python
'''
This code calculates the desired ratio of R2/R1 for 
a given boost in a circuit.

It uses the equation Vout = 1.25(1 + R2/R1)
'''



inputVoltage = 12 # This number cannot be changed for the formula
desiredVoltage = 24

desiredRatio = (desiredVoltage / 1.25) - 1 # This is a manipulated form of the provided formula "Vout = 1.25(1 + R2/R1)"
print(f'Desired R2/R1 ratio = {desiredRatio}')
```

```python
#desiredRatio = 18.2 # Uncomment if not using the above code

for i in range (len(commonResistors)):
    for j in range (len(commonResistors)):
        differences.append(differenceFinder(desiredRatio, commonResistors[i], commonResistors[j]))

bestThreeOptions = lowestThreeIndices(differences)

# For each of the three best indices, add their respective resistance indicies to two lists
# The first list contains R2 (numerator), and the second contains R1 (denominator)
for i in range(0, 3):
    bestResistorIndicesR2.append(int(bestThreeOptions[i] // len(commonResistors)))
    bestResistorIndicesR1.append(int(bestThreeOptions[i] % len(commonResistors)))

# Using the indices obtained from the above loop, create two new lists containing the actual resistance values
for i in range(0, 3):
    bestResistorsR2.append(commonResistors[bestResistorIndicesR2[i]])
    bestResistorsR1.append(commonResistors[bestResistorIndicesR1[i]])
```

```python
# Conclusions
print(f'The ideal ratio is {bestResistorsR2[0]}Ω / {bestResistorsR1[0]}Ω')
print(f'The second best is {bestResistorsR2[1]}Ω / {bestResistorsR1[1]}Ω')
print(f'And the third best is {bestResistorsR2[2]}Ω / {bestResistorsR1[2]}Ω')
```
